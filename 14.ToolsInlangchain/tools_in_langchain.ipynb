## Custom tool

```python
from langchain_core.tools import tool
```

```python
# Step 1- Create a function

def multiply(a, b):
    """Multiply two numbers """
    return a*b
```

```python
# Step 2- Add type hints

def multiply(a: int, b: int) -> int:
    """Multiply two numbers """
    return a*b
```

```python
# Step 3- Add tool Decorator

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers """
    return a*b
```

```python
result = multiply.invoke({'a':3, 'b':5})
```

```python
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

```python
print(multiply.args_schema.model_json_schema())  # This is what the llm sees when integrated with the tool
```
