```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# Create FAISS Vector store from documents
vectorstore = FAISS.from_documents(
    documents=documents,
    embedding=embeddings,
)
```

```python
# Enable MMR in the retriever

retriever = vectorstore.as_retriever(
    search_type = "mmr",
    search_kwargs={"k":3, "lambda_mult":1} # k= top reults, lambda_mult = relevance-diversity balannce , 1--> behaves as normal cosine similarity, 0--> gives diverse results
)
```

```python
query = "How can I improve API performance?"
results = retriever.invoke(query)
```

```python
print(results)
```

```python
for i, doc in enumerate(results):
    print(f"\n ------Results {i+1} ------")
    print(f"Content:\n {doc.page_content}")
```
